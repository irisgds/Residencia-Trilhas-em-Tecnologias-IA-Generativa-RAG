# Introdução à IA - Primeiro Contato com LLMs
Neste notebook, vamos fazer a nossa primeira chamada para a API da OpenAI utilizando as chaves configuradas com segurança.

### 1. Importação de Bibliotecas e Configuração
Primeiro, carregamos as variáveis de ambiente e iniciamos o cliente da OpenAI.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# Carrega as variáveis de ambiente do arquivo .env
load_dotenv()



client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url="https://openrouter.ai/api/v1"
)
```

### 2. Chamada à API
Usaremos o modelo configurado no `.env` (ou o `gpt-4o-mini` por padrão) para gerar um texto.

```python
# Usa o modelo definido no .env ou um valor padrão
modelo = os.getenv("OPENAI_MODEL", "gpt-4o-mini")

response = client.chat.completions.create(
    model=modelo,
    messages=[
        {"role": "user", "content": "write a haiku about ai"}
    ],
    store=True,
)

print(response.choices[0].message.content)
```
